```python
def oracle_syn3(tdd,anc_num=0,n_r=0):
    global S, node_qubit

    if abs(tdd.weight)<1e-10:
        cir = Circuit(n_r+1,[])
        return cir
    
    Q = [tdd.node]
    S=[]
    
    anc_num = 0
    n = tdd.node.key+1
    cir = Circuit(n+anc_num+1,[])
    anc_avi = n
    
    
    node_qubit = {}
    get_ref(tdd.node,[])
    
    
    # the_map=tdd.map
    # while the_map.level>-1:
    #     q = the_map.level
    #     if the_map.x==1:
    #         cir.data.append(Gate('x',{},q))
    #     the_map=the_map.father

    while len(Q)>0:
        u = Q.pop(0)
        S.append(u)
        the_map = u.out_maps[1]
        while the_map.level>-1:
            q1 = u.key
            q = the_map.level
            if not u in node_qubit:
                cir.data.append(Gate('x',{q1:1},q))
            else:
                cir.data.append(Gate('x',{node_qubit[u]:1,q1:1},q))
            the_map=the_map.father

        #进一步优化
        if u.successor[0]==u.successor[1]:
            u.successor[0].ref-=2
            if u.successor[0].ref==0 and u.successor[0].key>-1:
                Q.append(u.successor[0])
            if u.successor[0].key>-1:
                if not u.successor[0] in node_qubit:
                    if u.successor[0].ref>0 or abs(u.out_weight[1])<1e-10:
                        node_qubit[u.successor[0]] = anc_avi
                        anc_avi+=1
                        anc_num+=1
                    elif u in node_qubit:
                        node_qubit[u.successor[0]]=node_qubit[u]
                if u.successor[0] in node_qubit:
                    if abs(u.out_weight[1])<1e-10:
                        if u in node_qubit and node_qubit[u.successor[0]]!=node_qubit[u]:
                            cir.data.append(Gate('x',{node_qubit[u]:1,u.key:0},node_qubit[u.successor[0]]))
                        else:
                            cir.data.append(Gate('x',{u.key:0},node_qubit[u.successor[0]]))
                    else:
                        if u in node_qubit and node_qubit[u.successor[0]]!=node_qubit[u]:
                            cir.data.append(Gate('x',{node_qubit[u]:1},node_qubit[u.successor[0]]))
            continue
                
        u.successor[0].ref-=1
        if u.successor[0].ref==0 and u.successor[0].key>-1:
            Q.append(u.successor[0])
        if u.successor[0].key>-1:
            if not u.successor[0] in node_qubit:
                node_qubit[u.successor[0]] = anc_avi
                anc_avi+=1
                anc_num+=1
            if u in node_qubit:
                cir.data.append(Gate('x',{u.key:0,node_qubit[u]:1},node_qubit[u.successor[0]]))
            else:
                cir.data.append(Gate('x',{u.key:0},node_qubit[u.successor[0]]))

            
        u.successor[1].ref-=1
        if u.successor[1].ref==0 and u.successor[1].key>-1:
            Q.append(u.successor[1])
        if u.successor[1].key>-1:
            if not u.successor[1] in node_qubit:
                node_qubit[u.successor[1]] = anc_avi
                anc_avi+=1
                anc_num+=1
            if u in node_qubit:
                cir.data.append(Gate('x',{u.key:1,node_qubit[u]:1},node_qubit[u.successor[1]]))
            else:
                cir.data.append(Gate('x',{u.key:1},node_qubit[u.successor[1]]))

    
    data_q = n+anc_num
    cir.num_qubits = n+anc_num+1
    # print(cir)
    for k in range(len(S)-1,-1,-1):
        u=S[k]

        if u.successor[0].key==-1:
            if abs(u.out_weight[1])<1e-10:
                if u in node_qubit:
                    cir.data.append(Gate('x',{node_qubit[u]:1,u.key:0},data_q))
                else:
                    cir.data.append(Gate('x',{u.key:0},data_q))
            else:
                if u in node_qubit:
                    cir.data.append(Gate('x',{node_qubit[u]:1},data_q))
                else:
                    cir.data.append(Gate('x',{},data_q))
                    
        if u.successor[0]==u.successor[1] and u.successor[0].key>-1:
            if u.successor[0] in node_qubit:
                if abs(u.out_weight[1])<1e-10:
                    if u in node_qubit and node_qubit[u.successor[0]]!=node_qubit[u]:
                        cir.data.append(Gate('x',{node_qubit[u]:1,u.key:0},node_qubit[u.successor[0]]))
                    else:
                        cir.data.append(Gate('x',{u.key:0},node_qubit[u.successor[0]]))
                else:
                    if u in node_qubit and node_qubit[u.successor[0]]!=node_qubit[u]:
                        cir.data.append(Gate('x',{node_qubit[u]:1},node_qubit[u.successor[0]]))
        else:
            if u.successor[1].key>-1:
                if u in node_qubit:
                    cir.data.append(Gate('x',{u.key:1,node_qubit[u]:1},node_qubit[u.successor[1]]))
                else:
                    cir.data.append(Gate('x',{u.key:1},node_qubit[u.successor[1]]))
            if u.successor[0].key>-1:
                if u in node_qubit:
                    cir.data.append(Gate('x',{u.key:0,node_qubit[u]:1},node_qubit[u.successor[0]]))
                else:
                    cir.data.append(Gate('x',{u.key:0},node_qubit[u.successor[0]]))

        gate_op = []
        the_map = u.out_maps[1]
        while the_map.level>-1:
            q1 = u.key
            q = the_map.level
            if u in node_qubit:
                gate_op.append(Gate('x',{node_qubit[u]:1,q1:1},q))
                # cir.data.append(Gate('x',{node_qubit[u]:1,q1:1},q))
            else:
                gate_op.append(Gate('x',{q1:1},q))
                # cir.data.append(Gate('x',{q1:1},q))
            the_map=the_map.father
        gate_op.reverse()
        cir.data+=gate_op
    # the_map=tdd.map
    # while the_map.level>-1:
    #     q = the_map.level
    #     if the_map.x==1:
    #         cir.data.append(Gate('x',{},q))
    #     the_map=the_map.father
    reverse_q = []
    the_map=tdd.map
    while the_map.level>-1:
        q = the_map.level
        reverse_q.append(q)
        the_map=the_map.father 
    for g in cir.data:
        for k in g.q_c:
            if k in reverse_q:
                g.q_c[k] = 1-g.q_c[k]
    return cir,anc_num


def verify_success(cir,data,n,anc_num):
    # print(data)
    for k in range(2**n):
        input = format(k & ((1 << n) - 1), f'0{n}b')
        current_state = {n-1-q : int(input[q]) for q in range(n)}
        for n_num in range(anc_num):
            current_state[n+n_num] = 0
        current_state[n+anc_num] = 0
        current_state_s = copy.copy(current_state)
        for g in cir.data:
            flag = True
            for q in g.q_c:
                if g.q_c[q]!=current_state[q]:
                    flag = False
                    break
            if flag:
                current_state[g.q_t]=1-current_state[g.q_t]
        for q in range(len(current_state)-1):
            if current_state[q]!=current_state_s[q]:
                print('Not Successful, see input:',input)
                break                       
        if current_state[n+anc_num]!=data[k]:
            print('Not Successful, see input:',input)
    print('Successful')
```
